## Results non-gaussianity and B-HF for CKI in the Berillium isotopes

#### Imports

```python
import pickle as pkl
import matplotlib.pyplot as plt
import numpy as np
```

#### Data

```python
with open("data/results_gaussianity/results.pkl", "rb") as f:
    results = pkl.load(f)
```

#### Results

```python
#             "relative_errors_in_energy_bhf": relative_errors_in_energy_bhf,
#             "fidelities_bhf": fidelities_bhf,
#             "is_hartree_fock_energy": is_hartree_fock_energy,
#             "fidelities_bhf_gaussian": fidelities_bhf_gaussian,
#             "is_hartree_fock_gaussian": is_hartree_fock_gaussian,
#             "measure_from_relative_errors": measure_from_relative_errors,
#             "measure_from_fidelities": measure_from_fidelities,

print(len(results["relative_errors_in_energy_bhf"]))
plt.plot(
    np.arange(
        len(results["relative_errors_in_energy_bhf"]),
    ),
    results["relative_errors_in_energy_bhf"],
    label="Relative Errors in Energy BHF",
)
plt.scatter(
    np.arange(
        len(results["relative_errors_in_energy_bhf"]),
    ),
    results["relative_errors_in_energy_bhf"],
    label="Relative Errors in Energy BHF",
)
plt.xticks(np.arange(len(results["relative_errors_in_energy_bhf"])), results["labels"])
plt.xlabel("Isotopes")
plt.ylabel("Relative Errors in Energy HF")
plt.show()
```

```python
plt.plot(
    np.arange(len(results["fidelities_bhf_gaussian"])),
    results["fidelities_bhf_gaussian"],
)
plt.scatter(
    np.arange(len(results["fidelities_bhf_gaussian"])),
    results["fidelities_bhf_gaussian"],
    label="Fidelities Closest Gaussian state",
)

plt.plot(
    np.arange(len(results["fidelities_bhf"])),
    results["fidelities_bhf"],
)
plt.scatter(
    np.arange(len(results["fidelities_bhf"])),
    results["fidelities_bhf"],
    label="Fidelities BHF",
)
plt.legend(fontsize=10)
plt.xticks(
    np.arange(len(results["fidelities_bhf_gaussian"])), results["labels"], rotation=0
)
plt.xlabel("Isotopes")
plt.ylabel("Gaussianity")
plt.show()
```
